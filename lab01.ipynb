# API - Applcation Programming Interface

This module will focus on the introduction of APIs. An API (or Application Programming Interface) is a set of rules that allows different applications to communicate and exchange information. APIs define how one application can request data or services from others, and is an important part of the interoperability between services today. APIs have existed for decades and have evolved alongside modern computing. While early APIs were primarily used within operating systems and enterprise software, today's web APIs connect websites, mobile applications, cloud services, and Internet of Things (IoT) devices around the world.

APIs are a fundamental part of modern software. Every day, millions of applications will exchange data through APIs without users ever noticing. Whether you are checking the weather, signing in to a service using your Google or iCloud-account, paying for something with Vipps, an API is often working behind the scenes.

Throughout this course, further into your education and eventually in your work, you will be working with APIs, so understanding how they work will be an important part for future endeavours.


# Real-world examples
Have you ever signed in to an application using your Google account instead of creating a new account? Services such as Strava and ChatGPT allow you to do this through Google's authentication API. This allows you to sign in quickly while keeping your login credentials secure.

Weather apps, such as Yr, display forecasts from locations all around the world. Rather than storing all this information themselves, they request weather data from the weather service APIs. The application sends information such as a city name or GPS location, and the API returns the current weather and forecast for that location.

Another common example is online shopping. When shopping online, you should never have to enter your banking credentials directly into the store's website. Instead, the website communicates with trusted payment providers such as Vipps or PayPal through their APIs.
These APIs allow payment information to be transferred securely, so the online store never needs direct access to your banking details.


# How developers use APIs 

When a service provides an API, it also provides **API documentation**. The documentation (Docs) will explain how developers can communicate with the API, including which requests are available, what information must be provided, and what responses can be expected, including both the data returned and the format of the response.

Communication with an API follows a simple pattern: a client sends a request, and the API returns a response.

To communicate with an API, you typically: 
- Read the API docs.
- Send a request to a specific **API endpoint**, including any required **input parameters**.
- Authenticate using an **API key** or another authentication method, if required.
- Receive a response containing the requested information.

What makes it so easy to work with APIs is that you do not need to know how the service works internally. An API endpoint can be thought of as a door: you send a request through it, and the application returns a rresponse. The internal implementation remains hidden, allowing developers to use the service without understanding all of the code behind it.
